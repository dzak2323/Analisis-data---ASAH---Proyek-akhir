* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?

## Import Semua Packages/Library yang Digunakan

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style='darkgrid')
print("Library berhasil di-import!")
```

## Data Wrangling

### Gathering Data

#### Load df ...

```python
# Load dataset harian penyewaan sepeda
day_df = pd.read_csv("day.csv")

# Menampilkan 5 baris pertama untuk memastikan data terbaca
display(day_df.head())
```

**Insight:** 
Dataset `day.csv` berhasil dimuat dengan sempurna. Terdapat 731 baris data agregasi harian penyewaan sepeda yang mencakup rentang waktu tahun 2011 hingga 2012. Data ini memuat 16 kolom penting, termasuk metrik cuaca yang sudah dinormalisasi (suhu, kelembapan, kecepatan angin) serta jumlah transaksi (casual, registered, dan total/cnt).

### Assessing Data

#### Identifying ... problem

```python
# Cek info tipe data tiap kolom
print("Informasi Tipe Data:")
day_df.info()

# Cek jumlah missing value
print("\nJumlah Missing Value:")
print(day_df.isna().sum())

# Cek jumlah duplikat
print("\nJumlah Data Duplikat: ", day_df.duplicated().sum())
```

**Masalah yang Ditemukan:**
* 1. **Tipe Data Tidak Sesuai:** Kolom `dteday` (tanggal) memiliki tipe data `object` (string), padahal seharusnya `datetime`.
* 2. **Representasi Nilai Kurang Intuitif:** Nilai pada kolom `season` (musim) direpresentasikan dengan angka (1, 2, 3, 4). Ini akan menyulitkan   pembacaan data saat divisualisasikan.

**Langkah Pembersihan (Cleaning):**
* 1. Mengubah tipe data kolom `dteday` menjadi `datetime` menggunakan fungsi `pd.to_datetime()`.
* 2. Melakukan *mapping* pada kolom `season` menjadi label teks yang jelas ('Spring', 'Summer', 'Fall', 'Winter').

**Insight:** 
Berdasarkan proses identifikasi, kualitas dataset sudah sangat baik: tidak ditemukan adanya *missing value* (0 pada semua kolom) maupun data duplikat (0). Namun, ditemukan masalah pada format tipe data `dteday` yang masih berupa *object* (seharusnya *datetime*) dan kolom `season` yang masih menggunakan representasi angka (1, 2, 3, 4) sehingga kurang intuitif untuk tahapan visualisasi nanti.

### Cleaning Data

#### Fixing ... problem

```python
# 1. Memperbaiki tipe data dteday menjadi datetime
day_df['dteday'] = pd.to_datetime(day_df['dteday'])

# 2. Mapping kolom season dari angka menjadi string
season_mapping = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
day_df['season'] = day_df['season'].map(season_mapping)

# Cek hasil pembersihan
print("Info dataset setelah cleaning:")
day_df.info()

print("\n5 Baris pertama setelah label musim diubah:")
display(day_df[['dteday', 'season', 'temp', 'cnt']].head())
```

**Insight:** 
Masalah format data telah berhasil diselesaikan. Tipe data kolom `dteday` kini sudah menjadi `datetime64[ns]`. Nilai numerik pada kolom `season` juga telah dikonversi menjadi label teks kategorikal ('Spring', 'Summer', 'Fall', 'Winter'). Dataset kini sudah bersih dan siap untuk dieksplorasi.

## Exploratory Data Analysis (EDA)

### Eksplorasi Pertanyaan 1: Tren Penyewaan Berdasarkan Musim
Kita akan mengelompokkan data berdasarkan kolom `season` (musim) untuk melihat total (`sum`) dan rata-rata (`mean`) penyewaan sepeda harian.

```python
# Mengelompokkan data berdasarkan musim
season_summary = day_df.groupby('season')['cnt'].agg(['sum', 'mean', 'max', 'min']).reset_index()

# Mengurutkan berdasarkan rata-rata penyewaan tertinggi
season_summary = season_summary.sort_values(by='mean', ascending=False)

print("Rangkuman Penyewaan Sepeda Berdasarkan Musim:")
display(season_summary)
```

### Eksplorasi Pertanyaan 2: Segmentasi Suhu (Analisis Lanjutan - Binning)
Ini adalah penerapan teknik pengelompokan (*Clustering/Binning*) secara manual. Kita akan membagi kolom suhu (`temp`) ke dalam 3 rentang nilai yang sama besar: **Rendah**, **Sedang**, dan **Tinggi**, lalu melihat dampaknya terhadap rata-rata penyewaan sepeda.

```python
# Membuat kategori suhu (Binning) menjadi 3 kelompok
day_df['temp_group'] = pd.cut(day_df['temp'], bins=3, labels=['Rendah', 'Sedang', 'Tinggi'])

# Mengelompokkan berdasarkan kategori suhu untuk melihat rata-rata penyewaan
temp_summary = day_df.groupby('temp_group', observed=False)['cnt'].mean().reset_index()

# Mengubah nama kolom agar lebih informatif
temp_summary.rename(columns={'cnt': 'rata_rata_penyewaan'}, inplace=True)

# Mengurutkan dari rata-rata penyewaan tertinggi
temp_summary = temp_summary.sort_values(by='rata_rata_penyewaan', ascending=False)

print("Rata-rata Penyewaan Sepeda Berdasarkan Kategori Suhu:")
display(temp_summary)
```

**Insight:** 
Dari tabel pivot hasil pengelompokan data, rata-rata penyewaan sepeda memuncak pada Musim Gugur (*Fall*) dengan angka menembus 5.644 penyewaan/hari. Pada analisis lanjutan (Binning suhu udara), terlihat tren yang sejalan: suhu udara di kategori "Tinggi" menghasilkan rata-rata tertinggi sebesar 5.644 penyewaan/hari, sangat kontras dengan saat suhu udara "Rendah" yang hanya mencapai 2.569 penyewaan/hari.

## Visualization & Explanatory Analysis

### **Pertanyaan 1**: Bagaimana tren jumlah total penyewaan sepeda harian berdasarkan musim (season) sepanjang tahun 2011 hingga 2012?

```python
plt.figure(figsize=(10, 6))

# Membuat barplot dengan seaborn
sns.barplot(
    x='season', 
    y='mean', 
    data=season_summary.sort_values(by='mean', ascending=False),
    palette="Blues_d"
)

# Menambahkan judul dan label yang bersih
plt.title("Rata-rata Penyewaan Sepeda Harian Berdasarkan Musim", loc="center", fontsize=15, pad=15)
plt.xlabel("Musim", fontsize=12)
plt.ylabel("Rata-rata Penyewaan", fontsize=12)
plt.tick_params(axis='x', labelsize=11)
plt.tick_params(axis='y', labelsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

### **Pertanyaan 2**: Berdasarkan segmentasi suhu udara harian (temp) ke dalam 3 kategori (Rendah, Sedang, Tinggi), kategori suhu manakah yang menghasilkan rata-rata penyewaan sepeda terbanyak?

```python
plt.figure(figsize=(8, 5))

# Membuat barplot untuk kategori suhu
sns.barplot(
    x='temp_group', 
    y='rata_rata_penyewaan', 
    data=temp_summary,
    palette="YlOrBr_r"
)

plt.title("Rata-rata Penyewaan Sepeda Berdasarkan Kategori Suhu", loc="center", fontsize=15, pad=15)
plt.xlabel("Kategori Suhu (Rendah, Sedang, Tinggi)", fontsize=12)
plt.ylabel("Rata-rata Penyewaan", fontsize=12)
plt.tick_params(axis='x', labelsize=11)
plt.tick_params(axis='y', labelsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

**Insight:** 
Visualisasi bar chart secara jelas mengonfirmasi temuan EDA. Grafik pertama mendemonstrasikan pola musiman yang kuat, di mana *Fall* dan *Summer* mendominasi grafik penyewaan. Grafik kedua secara visual membuktikan adanya korelasi positif antara tingginya suhu udara (kategori Tinggi dan Sedang) terhadap lonjakan minat pelanggan dalam menyewa sepeda.
